```python
import kaggle as kg
from pathlib import Path
import os

PROJECT_ROOT = Path(os.getcwd()).parents[1]

print(PROJECT_ROOT)

raw_dataset_path = os.path.join(PROJECT_ROOT, 'data', 'raw')

print(raw_dataset_path)

kg.api.dataset_download_files("rupakroy/online-payments-fraud-detection-dataset", path=raw_dataset_path, unzip=True, quiet=False)
```

```python
import pandas as pd

raw_df = pd.read_csv(os.path.join(raw_dataset_path, "PS_20174392719_1491204439457_log.csv"))

print(len(raw_df))
print(raw_df.head(5))
```

```python
raw_df.dropna(inplace=True)
print(len(raw_df))
raw_df.drop_duplicates(inplace=True)
print(len(raw_df))
```
